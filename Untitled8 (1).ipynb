**TASK** **1**

```python
import json
import pandas as pd
import requests

# 1. Collect API Data
url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"
response = requests.get(url)
data = response.json()

# 2. Create DataFrame
df = pd.DataFrame(data["items"])

# 3. Select Required Columns
required_columns = [
    "name",
    "owner",
    "language",
    "stargazers_count",
    "forks_count",
    "watchers_count",
    "open_issues_count",
    "created_at",
    "updated_at",
    "license",
]
df = df[required_columns]

# 4. Extract Required Values (owner & license)
df["owner"] = df["owner"].apply(
    lambda x: x["login"] if isinstance(x, dict) and "login" in x else None
)
df["license"] = df["license"].apply(
    lambda x: x["name"] if isinstance(x, dict) and "name" in x else None
)

# 5. Handle Missing Values
df["language"] = df["language"].fillna("Unknown")
df["license"] = df["license"].fillna("No License")

# 6. Handle Duplicate Records
df = df.drop_duplicates()

# 7. Prepare the Dataset (Convert dates & rename columns)
df["created_at"] = pd.to_datetime(df["created_at"])
df["updated_at"] = pd.to_datetime(df["updated_at"])

df = df.rename(
    columns={
        "stargazers_count": "stars",
        "forks_count": "forks",
        "watchers_count": "watchers",
        "open_issues_count": "open_issues",
        "created_at": "created_date",
        "updated_at": "updated_date",
    }
)

# 8. Save and Verify the Dataset
df.to_csv("github_projects.csv", index=False)

# Verification step
reloaded_df = pd.read_csv("github_projects.csv")
print("Data saved and verified successfully! Shape:", reloaded_df.shape)
df.info()
df.head()
```

```python
import sqlite3
import matplotlib.pyplot as plt
import pandas as pd

# Step 1: Create SQLite Database and Import Data
conn = sqlite3.connect("github_data.db")
df = pd.read_csv("github_projects.csv")
df.to_sql("Repositories", conn, if_exists="replace", index=False)

# Step 2: Filtering & Searching Queries
q_stars = pd.read_sql_query(
    "SELECT * FROM Repositories WHERE stars > 10000", conn
)
q_search = pd.read_sql_query(
    "SELECT * FROM Repositories WHERE name LIKE '%Machine%'", conn
)

# Step 3: Logical Operators (AND, OR, NOT)
q_logic_1 = pd.read_sql_query(
    "SELECT * FROM Repositories WHERE stars > 5000 AND language = 'Python'",
    conn,
)
q_logic_2 = pd.read_sql_query(
    "SELECT * FROM Repositories WHERE language = 'C++' OR language = 'Java'",
    conn,
)
q_logic_3 = pd.read_sql_query(
    "SELECT * FROM Repositories WHERE NOT language = 'Python'", conn
)

# Step 4: Sorting & Limiting Queries (Top 10 Repositories)
top10_repos = pd.read_sql_query(
    "SELECT name, stars FROM Repositories ORDER BY stars DESC LIMIT 10", conn
)

# Step 5: Aggregate Functions
summary_stats = pd.read_sql_query(
    "SELECT COUNT(*) as total_repos, AVG(stars) as avg_stars FROM Repositories",
    conn,
)

# Step 6: Grouping Analysis (Languages > 5 Repositories)
grouped_languages = pd.read_sql_query(
    "SELECT language, COUNT(*) as repo_count FROM Repositories GROUP BY language HAVING repo_count > 5",
    conn,
)

# Step 7: Create Visualizations
# Chart 1: Top 10 Repositories by Stars
plt.figure(figsize=(10, 5))
plt.barh(top10_repos["name"], top10_repos["stars"], color="skyblue")
plt.xlabel("Number of Stars")
plt.ylabel("Repository Name")
plt.title("Top 10 Machine Learning Repositories by Stars")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig("top_10_repos.png")
plt.show()

# Chart 2: Creation Trends Over Time
df["created_year"] = pd.to_datetime(df["created_date"]).dt.year
yearly_counts = df["created_year"].value_counts().sort_index()

plt.figure(figsize=(8, 4))
plt.plot(
    yearly_counts.index, yearly_counts.values, marker="o", color="purple"
)
plt.title("Repository Creation Trend Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Repositories")
plt.grid(True)
plt.tight_layout()
plt.savefig("creation_trend.png")
plt.show()

conn.close()
```

## Ethics Reflection
* **Question 1: Why is it important to verify data collected from public APIs?**
  * **Answer:** Public API data can be incomplete, outdated, or duplicated. Verifying the data ensures its quality, prevents analytical bias, and leads to accurate results.

* **Question 2: Why should data analysts document the source of their data?**
  * **Answer:** Documenting data sources builds trust, maintains transparency, and allows other analysts to reproduce and verify the findings.

* **Question 3: How can missing or inaccurate data affect data analysis and decision-making?**
